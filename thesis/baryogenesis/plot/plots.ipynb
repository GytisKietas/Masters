```python
import argparse
import numpy as np
import matplotlib.pyplot as plt
import os
import logging
```

*LOAD DATA*

```python
FILE_DIR = "../code/SAVED_HERE/"
# FILE_DIR = "../code/CHECKPOINTS/"
FILE_NAME = "slowroll_test.csv"
CSV = FILE_DIR + FILE_NAME

data = np.genfromtxt(CSV, delimiter=",", names=True)
print(data)
```

*PLOT*

```python
plt.figure(figsize=(8, 5))

x = data["N"]
y = data["phi"]
# y_prime = data["phi_prime"]

plt.plot(x, y, label="numerical")

# plt.plot(x,y_prime, label="y prime")

plt.xlabel("x label")
plt.ylabel("y label")
# plt.yscale(value="log")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def phi_of_N(N, phi_i, kappa, b):
    return -b * np.log(
        np.exp(-phi_i / b) + (kappa / b**2) * N
    )

def phi_of_N_v2(N, phi_i, kappa, b):
    return -b * np.log((kappa / b**2) * N)
```

```python
print(x)
```

```python
analytic_phi = phi_of_N(x, 1.0, 1.0, 1.0)

# analytic_phi = analytic_phi + 1

plt.plot(x, analytic_phi, label="analytic")

plt.plot(x, y, label="numerical")

plt.title(r"Slowrol with $\kappa = 1$, $b=1.0$, $\phi_i = 1.0$, $N \in [0;60]$")
plt.xlabel(r"$N$")
plt.ylabel(r"$\phi$")
# plt.yscale(value="log")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
